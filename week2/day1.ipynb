## Và bây giờ cho vui — một cuộc hội thoại đối kháng giữa các Chatbots..

Bạn đã quen với việc tổ chức prompts thành lists như:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

Thực ra cấu trúc này có thể phản ánh lịch sử hội thoại dài hơn:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

Và chúng ta dùng cách này để tương tác dài hơn, có history (lịch sử).

````python
# Hãy tạo một cuộc hội thoại giữa GPT-4.1-mini và Claude-haiku-4.5
# Chúng ta dùng phiên bản rẻ của models nên chi phí sẽ rất nhỏ

gpt_model = "gpt-4.1-mini"
claude_model = "anthropic/claude-haiku-4.5"

gpt_system = "Bạn là một chatbot rất thích tranh cãi; \
bạn không đồng ý với bất cứ điều gì trong cuộc hội thoại và thách thức mọi thứ, theo cách snarky (châm chọc)."

claude_system = "Bạn là một chatbot rất lịch sự, nhã nhặn. Bạn cố đồng ý với \
mọi điều người kia nói, hoặc tìm common ground (điểm chung). Nếu người kia thích tranh cãi, \
bạn cố làm họ dịu lại và tiếp tục trò chuyện."

gpt_messages = ["Hi there"]
claude_messages = ["Hi"]
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

## Bài tập nâng cao: hội thoại 3 bên + Ollama

Ba vai, ba model:

| Vai | Tính cách | Model |
|---|---|---|
| **Alex** | Tranh cãi, snarky | GPT-4.1-mini (OpenAI) |
| **Blake** | Lịch sự, tìm điểm chung | Gemini 3.1 Flash Lite |
| **Charlie** | Trung lập, phân tích | llama3.2 (Ollama, local) |

Mỗi lần gọi chỉ gửi **1 system prompt + 1 user prompt**. User prompt chứa toàn bộ transcript có ghi tên người nói — vì API chỉ có `user` / `assistant`, không biểu diễn đủ 3 danh tính.

````python
TOPIC = "Có nên dùng model local (Ollama) thay hẳn cho frontier model khi học LLM Engineering?"

speakers = {
    "Alex": {
        "client": openai,
        "model": "gpt-4.1-mini",
        "max_tokens": 220,
        "system": """Bạn là Alex, chatbot rất thích tranh cãi; bạn không đồng ý với bất cứ điều gì
và thách thức mọi thứ, theo cách snarky (châm chọc).
Bạn đang trong cuộc hội thoại với Blake và Charlie.
Trả lời ngắn (2–4 câu), bằng tiếng Việt.
Chỉ viết đúng một lượt nói của Alex. Không ghi tên mình. Không giả lời Blake hay Charlie.""",
    },
    "Blake": {
        "client": gemini,
        "model": "gemini-3.1-flash-lite",
        "max_tokens": 220,
        "system": """Bạn là Blake, chatbot rất lịch sự, nhã nhặn. Bạn cố đồng ý hoặc tìm common ground (điểm chung).
Nếu người kia thích tranh cãi, bạn cố làm họ dịu lại.
Bạn đang trong cuộc hội thoại với Alex và Charlie.
Trả lời ngắn (2–4 câu), bằng tiếng Việt.
Chỉ viết đúng một lượt nói của Blake. Không ghi tên mình. Không giả lời Alex hay Charlie.""",
    },
    "Charlie": {
        "client": ollama,
        "model": "llama3.2",
        "max_tokens": 220,
        "system": """Bạn là Charlie, chatbot trung lập, thích phân tích.
Bạn chỉ ra điểm mạnh/yếu của cả hai phía và đề xuất cách kiểm chứng bằng thử nghiệm nhỏ.
Bạn đang trong cuộc hội thoại với Alex và Blake.
Trả lời ngắn (2–4 câu), bằng tiếng Việt.
Chỉ viết đúng một lượt nói của Charlie. Không ghi tên mình. Không giả lời Alex hay Blake.""",
    },
}

SPEAKER_ORDER = ["Alex", "Blake", "Charlie"]
````

````python
def format_conversation(history):
    if not history:
        return "(Chưa ai nói. Hãy mở đầu cuộc trò chuyện về chủ đề đã cho.)"
    return "\n".join(f"{turn['name']}: {turn['text']}" for turn in history)


def call_speaker(name, history, debug=False):
    speaker = speakers[name]
    others = [n for n in SPEAKER_ORDER if n != name]
    conversation = format_conversation(history)
    user_prompt = f"""Bạn là {name}, đang nói chuyện với {others[0]} và {others[1]}.
Chủ đề: {TOPIC}
Cuộc hội thoại đến nay như sau:
{conversation}
Với những gì trên, hãy nói điều bạn muốn nói tiếp theo, với tư cách {name}.
Chỉ trả về lời thoại của {name}. Không mô phỏng lời người khác."""

    messages = [
        {"role": "system", "content": speaker["system"]},
        {"role": "user", "content": user_prompt},
    ]
    if debug:
        print(f"--- messages gửi cho {name} ---")
        for m in messages:
            print(f"[{m['role']}]\n{m['content']}\n")

    response = speaker["client"].chat.completions.create(
        model=speaker["model"],
        messages=messages,
        max_tokens=speaker["max_tokens"],
    )
    return response.choices[0].message.content.strip()
````

````python
history = []
ROUNDS = 3

display(Markdown(f"**Chủ đề:** {TOPIC}\n"))

for round_index in range(ROUNDS):
    for name in SPEAKER_ORDER:
        # In messages của lượt đầu để thấy 1 system + 1 user chứa toàn bộ transcript
        debug = round_index == 0 and name == "Alex"
        reply = call_speaker(name, history, debug=debug)
        history.append({"name": name, "text": reply})
        display(Markdown(f"### {name} (`{speakers[name]['model']}`):\n{reply}\n"))
````
