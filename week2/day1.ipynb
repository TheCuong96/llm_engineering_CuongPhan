## Và bây giờ cho vui — một cuộc hội thoại đối kháng giữa các Chatbots..

Bạn đã quen với việc tổ chức prompts thành lists như:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

Thực ra cấu trúc này có thể phản ánh lịch sử hội thoại dài hơn:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

Và chúng ta dùng cách này để tương tác dài hơn, có history (lịch sử).

````python
# Hãy tạo một cuộc hội thoại giữa GPT-4.1-mini và Claude-haiku-4.5
# Chúng ta dùng phiên bản rẻ của models nên chi phí sẽ rất nhỏ

gpt_model = "gpt-4.1-mini"
claude_model = "claude-haiku-4-5"

gpt_system = "Bạn là một chatbot rất thích tranh cãi; \
bạn không đồng ý với bất cứ điều gì trong cuộc hội thoại và thách thức mọi thứ, theo cách snarky (châm chọc)."

claude_system = "Bạn là một chatbot rất lịch sự, nhã nhặn. Bạn cố đồng ý với \
mọi điều người kia nói, hoặc tìm common ground (điểm chung). Nếu người kia thích tranh cãi, \
bạn cố làm họ dịu lại và tiếp tục trò chuyện."

gpt_messages = ["Hi there"]
claude_messages = ["Hi"]
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_gpt()
````

````python
def call_claude():
    messages = [{"role": "system", "content": claude_system}]
    for gpt, claude_message in zip(gpt_messages, claude_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = anthropic.chat.completions.create(model=claude_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_claude()
````

````python
call_gpt()
````

````python
gpt_messages = ["Hi there"]
claude_messages = ["Hi"]

display(Markdown(f"### GPT:\n{gpt_messages[0]}\n"))
display(Markdown(f"### Claude:\n{claude_messages[0]}\n"))

for i in range(5):
    gpt_next = call_gpt()
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_messages.append(gpt_next)
    
    claude_next = call_claude()
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    claude_messages.append(claude_next)
````

# Các bài tập nâng cao hơn

Hãy thử tạo hội thoại 3 bên, có thể đưa Gemini vào cuộc! Một học viên đã làm xong — xem bản triển khai trong thư mục community-contributions.

Cách đáng tin cậy nhất là nghĩ hơi khác về prompts: mỗi lần chỉ 1 system prompt và 1 user prompt, và trong user prompt liệt kê toàn bộ cuộc hội thoại đến nay.

Đại loại như:

```python
system_prompt = """
Bạn là Alex, một chatbot rất thích tranh cãi; bạn không đồng ý với bất cứ điều gì trong cuộc hội thoại và thách thức mọi thứ, theo cách snarky (châm chọc).
Bạn đang trong cuộc hội thoại với Blake và Charlie.
"""

user_prompt = f"""
Bạn là Alex, đang nói chuyện với Blake và Charlie.
Cuộc hội thoại đến nay như sau:
{conversation}
Với những gì trên, hãy nói điều bạn muốn nói tiếp theo, với tư cách Alex.
"""
```

Hãy tự làm trước khi xem solutions (đáp án). Cách dễ nhất là dùng OpenAI python client để gọi model Gemini (xem example Gemini thứ 2 ở trên).
