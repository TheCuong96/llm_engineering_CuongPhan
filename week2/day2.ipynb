# Ngày Gradio!

Hôm nay chúng ta sẽ xây User Interfaces (giao diện người dùng) bằng framework Gradio cực kỳ đơn giản.

Chuẩn bị niềm vui!

Lưu ý: màn hình Gradio có thể hiện dark mode (chế độ tối) hoặc light mode (chế độ sáng) tùy cài đặt máy tính của bạn.

```python
# imports (nhập thư viện)

import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
import gradio as gr  # oh yeah! (tuyệt!)
```

```python
# Nạp environment variables (biến môi trường) từ file .env
# In prefix (tiền tố) của keys để hỗ trợ debug
# Bạn có thể chọn bất kỳ providers (nhà cung cấp) nào — hoặc toàn bộ Ollama

load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')
openrouter_api_key = os.getenv('OPENROUTER_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key tồn tại và bắt đầu bằng {openai_api_key[:8]}")
else:
    print("OpenAI API Key chưa được đặt")
    
if anthropic_api_key:
    print(f"Anthropic API Key tồn tại và bắt đầu bằng {anthropic_api_key[:7]}")
else:
    print("Anthropic API Key chưa được đặt")

if google_api_key:
    print(f"Google API Key tồn tại và bắt đầu bằng {google_api_key[:8]}")
else:
    print("Google API Key chưa được đặt")

if openrouter_api_key:
    print(f"OpenRouter API Key tồn tại và bắt đầu bằng {openrouter_api_key[:3]}")
else:
    print("OpenRouter API Key chưa được đặt")
```

```python
# Kết nối OpenAI, Anthropic, Google và OpenRouter
# Anthropic API hết credit (400) — Claude sẽ được gọi qua OpenRouter

openai = OpenAI()

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
openrouter_url = "https://openrouter.ai/api/v1"

anthropic = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
openrouter = OpenAI(api_key=openrouter_api_key, base_url=openrouter_url)
```

```python
# Bọc một lời gọi GPT-4.1-mini trong một function (hàm) đơn giản

system_message = "Bạn là một trợ lý hữu ích"

def message_gpt(prompt):
    messages = [{"role": "system", "content": system_message}, {"role": "user", "content": prompt}]
    response = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)
    return response.choices[0].message.content
```

```python
# Có thể tiết lộ "training cut off" (mốc cắt dữ liệu huấn luyện), tức ngày gần nhất trong dữ liệu training

message_gpt("Hôm nay là ngày nào?")
```

## Đến lúc làm User Interface (giao diện người dùng)!

```python
# Một function đơn giản

def shout(text):
    print(f"Shout đã được gọi với input {text}")
    return text.upper()
```

## Ép dark mode (chế độ tối)

Gradio hiện light mode hoặc dark mode tùy cài đặt trình duyệt và máy tính. Có cách ép Gradio hiện dark mode, nhưng Gradio khuyên không nên — đó nên là tùy chọn của user (đặc biệt vì accessibility / khả năng tiếp cận). Nếu vẫn muốn ép dark mode cho màn hình của bạn, cách làm nằm bên dưới.

```python
# Định nghĩa biến này rồi truyền js=force_dark_mode khi tạo Interface

force_dark_mode = """
function refresh() {
    const url = new URL(window.location);
    if (url.searchParams.get('__theme') !== 'dark') {
        url.searchParams.set('__theme', 'dark');
        window.location.href = url.href;
    }
}
"""
gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never", js=force_dark_mode).launch()
force_dark_mode = """
function refresh() {
    const url = new URL(window.location);
    if (url.searchParams.get('__theme') !== 'dark') {
        url.searchParams.set('__theme', 'dark');
        window.location.href = url.href;
    }
}
"""
gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never", js=force_dark_mode).launch()
```

```python
# Dùng Markdown
# Bạn có thắc mắc vì sao vẫn set system_message dù code bên dưới không nhắc tới nó?
# Tôi đang tận dụng system_message là biến global (toàn cục), được dùng trong function message_gpt (hãy xem lại)
# Không phải software engineering hay ho, nhưng khá phổ biến khi R&D trên Jupyter Lab!

system_message = "Bạn là một trợ lý hữu ích, trả lời bằng markdown, không dùng code blocks"

message_input = gr.Textbox(label="Tin nhắn của bạn:", info="Nhập tin nhắn cho GPT-4.1-mini", lines=7)
message_output = gr.Markdown(label="Phản hồi:")

view = gr.Interface(
    fn=message_gpt,
    title="GPT", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Giải thích kiến trúc Transformer cho người không chuyên",
        "Giải thích kiến trúc Transformer cho kỹ sư AI đang học",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
# Tạo lời gọi stream (luồng) kết quả về dần
# Nếu muốn ôn Generators (từ khóa "yield"),
# hãy xem Intermediate Python guide trong thư mục guides

def stream_gpt(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = openai.chat.completions.create(
        model='gpt-4.1-mini',
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
def stream_gpt(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = openai.chat.completions.create(
        model='gpt-4.1-mini',
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input = gr.Textbox(label="Tin nhắn của bạn:", info="Nhập tin nhắn cho GPT-4.1-mini", lines=7)
message_output = gr.Markdown(label="Phản hồi:")

view = gr.Interface(
    fn=stream_gpt,
    title="GPT", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Giải thích kiến trúc Transformer cho người không chuyên",
        "Giải thích kiến trúc Transformer cho kỹ sư AI đang học",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
# Stream kết quả từ Claude, cùng kiểu generator như stream_gpt
# Anthropic API hết credit (400). Gọi Claude qua OpenRouter và giới hạn max_tokens
# vì OpenRouter tính credit theo trần token, không phải số token thực tế.

def stream_claude(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = openrouter.chat.completions.create(
        model="anthropic/claude-sonnet-4.5",
        messages=messages,
        stream=True,
        max_tokens=800,
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input = gr.Textbox(label="Tin nhắn của bạn:", info="Nhập tin nhắn cho Claude 4.5 Sonnet", lines=7)
message_output = gr.Markdown(label="Phản hồi:")

view = gr.Interface(
    fn=stream_claude,
    title="Claude", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Giải thích kiến trúc Transformer cho người không chuyên",
        "Giải thích kiến trúc Transformer cho kỹ sư AI đang học",
        ], 
    flagging_mode="never"
    )
view.launch()
```

## Bây giờ làm cho fancy (bắt mắt) hơn

Nhớ xem Intermediate Python Guide nếu chưa chắc về generators và "yield"

```python
def stream_model(prompt, model):
    if model=="GPT":
        result = stream_gpt(prompt)
    elif model=="Claude":
        result = stream_claude(prompt)
    else:
        raise ValueError("Unknown model")
    yield from result
```

```python
message_input = gr.Textbox(label="Tin nhắn của bạn:", info="Nhập tin nhắn cho LLM", lines=7)
model_selector = gr.Dropdown(["GPT", "Claude"], label="Chọn model", value="GPT")
message_output = gr.Markdown(label="Phản hồi:")

view = gr.Interface(
    fn=stream_model,
    title="LLMs", 
    inputs=[message_input, model_selector], 
    outputs=[message_output], 
    examples=[
            ["Giải thích kiến trúc Transformer cho người không chuyên", "GPT"],
            ["Giải thích kiến trúc Transformer cho kỹ sư AI đang học", "Claude"]
        ], 
    flagging_mode="never"
    )
view.launch()
```
