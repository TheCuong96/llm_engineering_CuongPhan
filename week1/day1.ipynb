## Các loại prompt (lời nhắc)

Bạn có thể đã biết rồi — nhưng nếu chưa, bạn sẽ sớm rất quen thuộc!

Các model như GPT đã được huấn luyện để nhận instruction (hướng dẫn) theo một cách cụ thể.

Chúng kỳ vọng nhận:

**A system prompt (prompt hệ thống)** — cho chúng biết đang thực hiện nhiệm vụ gì và nên dùng giọng điệu nào

**A user prompt (prompt người dùng)** — phần mở đầu hội thoại mà chúng cần trả lời

````python
# Định nghĩa system prompt (prompt hệ thống) của chúng ta — sau này bạn có thể thử nghiệm, đổi câu cuối thành 'Respond in markdown in Spanish.'

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# Định nghĩa user prompt (prompt người dùng) của chúng ta

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## Messages (tin nhắn)

API của OpenAI kỳ vọng nhận messages theo một cấu trúc cụ thể.
Nhiều API khác cũng dùng chung cấu trúc này:

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```
Để bạn xem trước, 2 ô tiếp theo thực hiện một lời gọi khá đơn giản — chúng ta chưa “vắt sức” GPT mạnh mẽ đâu (chưa!)

````python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]

response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
response.choices[0].message.content
````

## Và bây giờ hãy xây các messages hữu ích cho GPT-4.1-mini, bằng một function (hàm)

````python
# Xem function (hàm) này tạo đúng định dạng như trên như thế nào

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# Thử cái này, rồi thử thêm vài website nữa

messages_for(ed)
````

## Đến lúc ghép mọi thứ lại — API của OpenAI rất đơn giản!

````python
# Và bây giờ: gọi OpenAI API. Bạn sẽ rất quen với việc này!

def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
````

````python
summarize("https://edwarddonner.com")
````

````python
# Một function (hàm) để hiển thị kết quả đẹp trong output, dùng markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_summary("https://edwarddonner.com")
````

Nếu muốn thêm thay đổi của bạn vào thư mục đó, hãy gửi một Pull Request (yêu cầu hợp nhất, viết tắt PR) với phiên bản mới trong thư mục đó, rồi tôi sẽ merge (hợp nhất) thay đổi của bạn.

Nếu bạn chưa thành thạo git (và tôi cũng vậy!) thì tôi đã ghi hướng dẫn đầy đủ trong thư mục guides, guide 3, và dán lại ở đây:

Đây là các bước tổng quát khi tạo PR và hướng dẫn chính:  
https://edwarddonner.com/pr  

Hãy kiểm tra trước khi gửi:  
1. PR của bạn chỉ chứa thay đổi trong community-contributions (trừ khi chúng ta đã bàn)  
2. Tất cả output (kết quả) của notebook đều đã được xóa sạch  
3. Tổng cộng dưới 2.000 dòng code, và không quá nhiều file  
4. Đừng kèm file test không cần thiết, README quá dài dòng, `.env.example`, emoji, hay các artifact (mẩu nội dung) do LLM tạo ra!

Cảm ơn rất nhiều!
