### NHƯNG TRƯỚC HẾT — Một phần thú vị: xem "model inference" (suy luận mô hình) hoạt động thế nào qua OpenAI

```python
from visualizer import TokenPredictor, create_token_graph, visualize_predictions

# Prompt: mô tả màu cam cho người chưa từng nhìn thấy. Đổi màu (orange → blue, ...) để xem token khác đi.
# message = "In one sentence, describe the color orange to someone who has never been able to see"
message = "Hãy mô tả màu xanh dương trong một câu cho một người chưa từng nhìn thấy."
model_name = "gpt-4.1-mini"

predictor = TokenPredictor(model_name)
predictions = predictor.predict_tokens(message)  # dự đoán token tiếp theo từng bước
G = create_token_graph(model_name, predictions)  # tạo đồ thị xác suất token
plt = visualize_predictions(G)  # trực quan hóa quá trình inference
plt.show()
```
